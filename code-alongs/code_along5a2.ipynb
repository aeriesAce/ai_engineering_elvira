```python
class Person:
    def __init__(self, name: str, age: int, gender: str):
        if not isinstance(name, str):
            raise TypeError(f"Name must be of type str not {type(name)}")
        
        self.name = name
        self.age = age
        self.gender = gender

    @property
    def age(self):
        return self._age
    @age.setter
    def age(self, value: int):
        if not isinstance(value, int):
            raise TypeError(f"age must be of type int not {type(age)} that you have provided")
        if value < 0 or value > 125:
            raise ValueError(f"age must be between 0 and 125, not {value} that you provided")

        self._age = value

person5 = Person(name = "Sunset", age = 3, gender = "F")
person5.age
```

```python
try:
    Person(name = "Sunset", age = -3, gender = "F")
except ValueError as err:
    print(err)
```

```python
from pydantic import BaseModel

class Person(BaseModel):
    name: str
    age: int
    gender: str

person6 = Person(name = "Robert", age = 70, gender = "M")
person6
```

```python
person6.name, person6.age
```

```python
from pydantic import ValidationError

try:
    Person(name = 55, age = -5, gender = 1.55)
except ValidationError as err:
    print(err)
```

```python
try:
    Person(name = "Sofia", gender=2.714, age = -5)
except ValidationError as err:
    print(err)
```

```python
Person(name = "Sofia", gender="F", age = -5)
```

```python
from pydantic import Field

class Person(BaseModel):
    name: str
    age: int = Field(gt = -1, lt=126)
    gender: str

try:
    Person(name = "Ace", age = 126, gender = "M")
except ValidationError as err:
    print(err)
```

```python
from typing import Literal

class Person(BaseModel):
    name: str
    age: int = Field(gt = -1, lt = 126)
    gender: Literal["M", "F"]

try:
    Person(name = "Ace", age = 5, gender = "Male")
except ValidationError as err:
    print(err)
```
