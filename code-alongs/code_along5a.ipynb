```python
import requests
from pydantic import BaseModel, computed_field
from pydantic import ValidationError

data = requests.get("https://pokeapi.co/api/v2/pokemon?limit=20").json()
data.keys()
```

```python
data["count"]
```

```python
data['results'][:5]
```

```python
class PokemonListResponse(BaseModel):
        results: list[dict]
        count: int

        @computed_field
        @property
        def number_stored_pokemons(self) -> int:
            return len(self.results)

pokemons = PokemonListResponse.model_validate(data)
pokemons
```

```python
pokemons.count
```

```python
pokemons.results[:5]
```

```python
pokemons.number_stored_pokemons
```

```python
pika_data = requests.get("https://pokeapi.co/api/v2/pokemon-species/pikachu").json()
pika_data.keys()
```

```python
class PokemonListResponse(BaseModel):
    results: list[dict]
    count: str

try:
    pokemon = PokemonListResponse.model_validate(pika_data)
except ValidationError as e:
    print(e)
```

```python
for key, values in pika_data.items():
    print(f"{key: <30} {type(values)}")
```

```python
class Pokemon(BaseModel):
    id: int
    name: str
    base_happiness: int
    evolution_chain: dict
    gender_rate: int
    generation: dict

pikachu = Pokemon.model_validate(pika_data)
pikachu
```

```python
try:
    pikachu.abilites
except AttributeError as err:
    print(err)
```

```python
serialized_pikachu = pikachu.model_dump_json()
serialized_pikachu
```
